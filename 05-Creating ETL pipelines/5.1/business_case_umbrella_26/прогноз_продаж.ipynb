## Прогнозирование
Температура для Варианта 14 (Варшава, средняя по рабочим дням) составляет примерно 8.3°C.

```python
import joblib
import pandas as pd

# Загрузка модели
# Убедитесь, что версия scikit-learn совпадает с той, на которой обучалась модель в Airflow
model = joblib.load("ml_model.pkl")

# Введите температуру из вашего варианта - Варшава, средняя t по раб. дням = 8.3
my_temp = 8.3

# Формирование датафрейма для предикта
input_data = pd.DataFrame({'temperature': [my_temp]})

# Предсказание объема продаж
prediction = model.predict(input_data)

print(f"При температуре {my_temp}°C прогнозируемые продажи составят: {prediction[0]:.2f}")
```
